```python
revenue_skew = df["Revenue"].skew()
print(revenue_skew)
```

```python
if revenue_skew > 0:
  print("Positive Skew")
elif revenue_skew < 0:
  print("Negative Skew")
else:
  print("Symmetric Distribution")
```

```python
units_skew = df["Units_Sold"].skew()
print(units_skew)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df["Revenue"], bins=20)
plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(df["Units_Sold"], bins=20,color="green",edgecolor="blue")
plt.title("Units Sold Distribution for Retail Dataset",fontweight="bold",color="red")
plt.xlabel("Units Sold",fontweight="bold",color="red")
plt.ylabel("Frequency")
plt.figure(figsize=(200,100))
plt.grid(axis='y',linestyle="--",color="black",alpha=0.4)
plt.show()
```

```python
summary = pd.DataFrame({
    "Metric":["Revenue","Units Sold"],
    "Skewness":[revenue_skew,units_skew]
})
summary
```
